```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}
```

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """
    cola = [[inicio]]  # cola de caminos, empieza con el estado inicial
    visitados = set()

    while cola:
        camino = cola.pop(0)  # sacar el primer camino de la cola (FIFO)
        nodo_actual = camino[-1]

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                # COMPLETAR: crear un nuevo camino que sea "camino + [vecino]"
                # y agregarlo a la cola
                
                nuevo_camino = camino + [vecino]
                cola.append(nuevo_camino)

    return None  # no se encontro un camino

resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```

```python
## BFS: buscando el camino más corto

La función usa una cola (una lista donde siempre saco el primer elemento con `.pop(0)`).
Cada elemento de la cola no es un nodo suelto, sino el camino completo recorrido hasta
ahí, así cuando encuentro el objetivo ya tengo la ruta completa, no solo el destino.

En cada vuelta del while, saco el camino más antiguo de la cola, me fijo en el último nodo
de ese camino, y si es el objetivo, listo, esa es la respuesta. Si no, reviso sus vecinos
y agrego al final de la cola una versión extendida del camino por cada vecino nuevo.

Como siempre saco de la cola por el principio, exploro primero todos los caminos más
cortos antes de pasar a los más largos, y por eso BFS siempre encuentra el camino con
menos pasos.

Resultado obtenido: Almacen -> Centro -> ZonaNorte (2 pasos), que coincide con lo visto
en la traza manual del algoritmo.
```

```python
import numpy as np

obs1 = np.array([22.5, 60])  # temperatura, humedad
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])

dataset = np.array([obs1, obs2, obs3, obs4])
print(dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)
```

```python
print("Primera observacion (fila 0):", dataset[0])
print("Todas las temperaturas (columna 0):", dataset[:, 0])
```

```python
producto = np.dot(obs1, obs2)
print("Producto escalar entre obs1 y obs2:", producto)
```

```python
# Formula manual
distancia_manual = np.sqrt(np.sum((obs1 - obs2) ** 2))
print("Distancia manual:", distancia_manual)

# Con la funcion de NumPy
distancia_numpy = np.linalg.norm(obs1 - obs2)
print("Distancia con NumPy:", distancia_numpy)
```

```python
## Vectores, matrices y distancia euclidiana

Cada observación (temperatura, humedad) la representé como un vector de NumPy, y
al juntar las cuatro se forma una matriz de 4x2. dataset[0] me da una fila
completa y dataset[:, 0] me da solo la columna de temperaturas.

El producto escalar (np.dot) multiplica y suma componente por componente entre
dos vectores.

La distancia euclidiana la calculé de dos formas (manual y con np.linalg.norm) y
dieron el mismo resultado. Mientras menor es la distancia, más parecidas son las
observaciones, que es justo la idea detrás de algoritmos como KNN.
```

```python
def dfs(grafo, inicio, objetivo):
    """
    Busca un camino entre 'inicio' y 'objetivo' usando DFS.
    """
    pila = [[inicio]]
    visitados = set()

    while pila:
        camino = pila.pop()  # sacar el ULTIMO camino de la pila (LIFO)
        nodo_actual = camino[-1]

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                # COMPLETADO
                nuevo_camino = camino + [vecino]
                pila.append(nuevo_camino)

    return None

resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (DFS):", resultado_dfs)
```

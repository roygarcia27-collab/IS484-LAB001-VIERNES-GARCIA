```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}
```

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """
    cola = [[inicio]]  # cola de caminos, empieza con el estado inicial
    visitados = set()

    while cola:
        camino = cola.pop(0)  # sacar el primer camino de la cola (FIFO)
        nodo_actual = camino[-1]

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                # COMPLETAR: crear un nuevo camino que sea "camino + [vecino]"
                # y agregarlo a la cola
                
                nuevo_camino = camino + [vecino]
                cola.append(nuevo_camino)

    return None  # no se encontro un camino

resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```

```python
## BFS: buscando el camino más corto

La función usa una cola (una lista donde siempre saco el primer elemento con `.pop(0)`).
Cada elemento de la cola no es un nodo suelto, sino el camino completo recorrido hasta
ahí, así cuando encuentro el objetivo ya tengo la ruta completa, no solo el destino.

En cada vuelta del while, saco el camino más antiguo de la cola, me fijo en el último nodo
de ese camino, y si es el objetivo, listo, esa es la respuesta. Si no, reviso sus vecinos
y agrego al final de la cola una versión extendida del camino por cada vecino nuevo.

Como siempre saco de la cola por el principio, exploro primero todos los caminos más
cortos antes de pasar a los más largos, y por eso BFS siempre encuentra el camino con
menos pasos.

Resultado obtenido: Almacen -> Centro -> ZonaNorte (2 pasos), que coincide con lo visto
en la traza manual del algoritmo.
```
